```python
#Ejercicio 1
ventas = {
    "productos":[
        {"fecha": "2024-01-01", "nombre": "Bolsas de Aire", "cantidad": 100, "precio": 19.99},
        {"fecha": "2026-11-14", "nombre": "Vaso de petroleo", "cantidad": 100, "precio": 999.99},
        {"fecha": "2025-12-08", "nombre": "Airbus A350-1000ULR", "cantidad": 12, "precio": 360.99}
    ]
}
# Calculo de ingresos totales (Ejercicio 2)
ganancias = 0
i = 0
while i < len(ventas["productos"]):
  producto = ventas["productos"][i]
  ganancias += (producto["cantidad"] * producto["precio"])
  i += 1

print(f"Se ha ganado ${ganancias:,}")


#Ejercicio 3
ventas_por_producto = [
        {"nombre": "Bolsas de Aire", "cantidad": 120},
        {"nombre": "Vaso de petroleo", "cantidad": 100},
        {"nombre": "Airbus A350-1000ULR", "cantidad": 12}

]

max_producto = max(ventas_por_producto, key=lambda item: item['cantidad'])
print(f"\nEl producto que mas se vendio fue {max_producto['nombre']}  {max_producto['cantidad']} ")

# Ejercicio 4

precios_por_producto = {}

for venta in ventas["productos"]:
    nombre = venta["nombre"]
    cantidad = venta["cantidad"]
    precio = venta["precio"]
    ingreso_total_por_venta = cantidad * precio

    if nombre in precios_por_producto:
        suma_ingresos_actual = precios_por_producto[nombre][0]
        cantidad_total_actual = precios_por_producto[nombre][1]
        precios_por_producto[nombre] = (suma_ingresos_actual + ingreso_total_por_venta, cantidad_total_actual + cantidad)
    else:
        precios_por_producto[nombre] = (ingreso_total_por_venta, cantidad)

print("\nPrecios promedio de venta por producto:")
for producto_nombre, (ingresos_totales, cantidad_unidades_vendidas) in precios_por_producto.items():
    if cantidad_unidades_vendidas > 0:
        precio_promedio = ingresos_totales / cantidad_unidades_vendidas
        print(f"- {producto_nombre}: {precio_promedio:.2f}")
    else:
        print(f"- {producto_nombre}: No se puede calcular el promedio (cero unidades vendidas).")

#Ejercico 5
ventas = {
    "productos":[
        {"fecha": "2024-01-01", "nombre": "Bolsas de Aire", "cantidad": 100, "precio": 19.99},
        {"fecha": "2026-11-14", "nombre": "Vaso de petroleo", "cantidad": 100, "precio": 999.99},
        {"fecha": "2025-12-08", "nombre": "Airbus A350-1000ULR", "cantidad": 12, "precio": 360.99} # Son 365 Millones de Dolares
    ]
}


ingresos_por_dia = {}


for venta in ventas["productos"]:
    fecha = venta["fecha"]
    ingreso_venta = venta["cantidad"] * venta["precio"]

    if fecha in ingresos_por_dia:
        ingresos_por_dia[fecha] += ingreso_venta
    else:
        ingresos_por_dia[fecha] = ingreso_venta


print("\nIngresos totales por día:")
for fecha, total_ingresos in ingresos_por_dia.items():
    print(f"- Fecha: {fecha}, Ingresos: {total_ingresos:.2f}")

#Ejercicio 6
resumen_ventas = {}

for producto_nombre, (ingresos_totales, cantidad_unidades_vendidas) in precios_por_producto.items():
    precio_promedio = 0
    if cantidad_unidades_vendidas > 0:
        precio_promedio = ingresos_totales / cantidad_unidades_vendidas

    resumen_ventas[producto_nombre] = {
        "cantidad_total": cantidad_unidades_vendidas,
        "ingresos_totales": ingresos_totales,
        "precio_promedio": precio_promedio
    }

print("\nResumen de Ventas por Producto:")
for producto_nombre, datos in resumen_ventas.items():
    print(f"- {producto_nombre}:")
    print(f"  Cantidad Total: {datos['cantidad_total']}")
    print(f"  Ingresos Totales: {datos['ingresos_totales']:.2f}")
    print(f"  Precio Promedio: {datos['precio_promedio']:.2f}")
```
